Desarrollar un sistema de análisis exploratorio, estadístico y predictivo sobre la base de datos de accidentes e incidentes aéreos de la NTSB (National Transport Safety Board) de los Estados unidos, aplicando técnicas de programación, métodos numéricos y aprendizaje automático para identificar patrones que apoyen la comprensión y la toma de decisiones en seguridad operacional.

**Bloque 12**

Variables categóricas

```python
# ==============================================================================
# 12. AUDITORÍA DE VARIABLES CATEGÓRICAS
# ==============================================================================

# Seleccionar variables categóricas (tipo object)
columnas_categoricas = df.select_dtypes(include=["object"]).columns

# Lista para almacenar los resultados
resumen_categoricas = []

for columna in columnas_categoricas:

    # Número de categorías distintas
    categorias = df[columna].nunique(dropna=True)

    # Moda (valor más frecuente)
    moda = df[columna].mode(dropna=True)

    if len(moda) > 0:
        valor_mas_frecuente = moda.iloc[0]
        frecuencia = df[columna].value_counts(dropna=True).iloc[0]
        porcentaje = (frecuencia / len(df)) * 100
    else:
        valor_mas_frecuente = None
        frecuencia = 0
        porcentaje = 0

    resumen_categoricas.append([
        columna,
        categorias,
        valor_mas_frecuente,
        frecuencia,
        round(porcentaje, 2)
    ])

# Crear DataFrame
tabla_categoricas = pd.DataFrame(
    resumen_categoricas,
    columns=[
        "Variable",
        "Categorías",
        "Valor más frecuente",
        "Frecuencia",
        "Porcentaje (%)"
    ]
)

# Ordenar por número de categorías
tabla_categoricas = tabla_categoricas.sort_values(
    by="Categorías",
    ascending=False
)

print("=" * 80)
print("AUDITORÍA DE VARIABLES CATEGÓRICAS")
print("=" * 80)

display(tabla_categoricas)
```

```python
# ==============================================================================
# 15.1 DEPURACIÓN DE VARIABLES TEMPORALES
# ==============================================================================

# Conversión de la fecha
df_clean["EventDate"] = pd.to_datetime(
    df_clean["EventDate"],
    errors="coerce",
    utc=True
)

# Variables temporales para análisis
df_clean["Year"] = df_clean["EventDate"].dt.year
df_clean["Month"] = df_clean["EventDate"].dt.month
df_clean["MonthName"] = df_clean["EventDate"].dt.month_name()
df_clean["Day"] = df_clean["EventDate"].dt.day
df_clean["Hour"] = df_clean["EventDate"].dt.hour
df_clean["WeekDay"] = df_clean["EventDate"].dt.day_name()
```

```python
# ==============================================================================
# 15.2 DEPURACIÓN DE VARIABLES AERONÁUTICAS
# ==============================================================================

# Variables de texto
variables_aeronauticas = [
    "Make",
    "Model",
    "AirCraftCategory",
    "AirportID",
    "AirportName"
]

for variable in variables_aeronauticas:

    df_clean[variable] = (
        df_clean[variable]
        .astype("string")
        .str.strip()
        .str.upper()
    )

# Número de motores
df_clean["NumberOfEngines"] = pd.to_numeric(
    df_clean["NumberOfEngines"],
    errors="coerce"
)
```

```python
# ==============================================================================
# 15.3 DEPURACIÓN DE VARIABLES OPERACIONALES
# ==============================================================================

variables_operacionales = [
    "PurposeOfFlight",
    "FAR",
    "AirCraftDamage",
    "WeatherCondition",
    "AmateurBuilt",
    "Operator"
]

for variable in variables_operacionales:

    df_clean[variable] = (
        df_clean[variable]
        .astype("string")
        .str.strip()
        .str.upper()
    )

# Valores vacíos como nulos
df_clean.replace("", np.nan, inplace=True)
```

```python
# ==============================================================================
# 15.4 DEPURACIÓN DE VARIABLES DE LESIONES
# ==============================================================================

# Nivel de lesión
df_clean["HighestInjuryLevel"] = (
    df_clean["HighestInjuryLevel"]
    .astype("string")
    .str.strip()
    .str.title()
)

# Variables numéricas de lesiones
variables_lesiones = [
    "FatalInjuryCount",
    "SeriousInjuryCount",
    "MinorInjuryCount"
]

for variable in variables_lesiones:

    df_clean[variable] = (
        pd.to_numeric(df_clean[variable], errors="coerce")
        .fillna(0)
        .astype(int)
    )
```

```python
# ==============================================================================
# 15.5 DEPURACIÓN DE LA VARIABLE ProbableCause
# ==============================================================================

df_clean["ProbableCause"] = (
    df_clean["ProbableCause"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)
```

```python
# ==============================================================================
# 15.6 DEPURACIÓN DE VARIABLES OPERACIONALES
# ==============================================================================

# ------------------------------------------------------------------------------
# PurposeOfFlight
# ------------------------------------------------------------------------------

df_clean["PurposeOfFlight"] = (
    df_clean["PurposeOfFlight"]
    .fillna("UNKNOWN")
    .replace(",", "UNKNOWN")
    .str.split(",")
    .str[0]
    .str.strip()
    .replace("", "UNKNOWN")
)

# ------------------------------------------------------------------------------
# AirCraftDamage
# ------------------------------------------------------------------------------

df_clean["AirCraftDamage"] = (
    df_clean["AirCraftDamage"]
    .fillna("UNKNOWN")
    .replace(",", "UNKNOWN")
    .str.split(",")
    .str[0]
    .str.strip()
    .replace("", "UNKNOWN")
)

# ------------------------------------------------------------------------------
# FAR
# ------------------------------------------------------------------------------

top_far = df_clean["FAR"].value_counts().head(10).index

df_clean["FAR"] = df_clean["FAR"].where(
    df_clean["FAR"].isin(top_far),
    "OTHER"
)

# ==============================================================================
# DEPURACIÓN DE WeatherCondition
# ==============================================================================

df_clean["WeatherCondition"] = (
    df_clean["WeatherCondition"]
    .fillna("UNKNOWN")
    .astype(str)
    .str.strip()
    .replace({
        "VFR": "VMC",
        "IFR": "IMC",
        "UNK": "UNKNOWN"
    })
)

# ==============================================================================
# DEPURACIÓN DE PurposeOfFlight
# ==============================================================================

df_clean["PurposeOfFlight"] = (
    df_clean["PurposeOfFlight"]
    .fillna("UNKNOWN")
    .astype(str)
    .str.strip()
    .replace({
        "UNK": "UNKNOWN"
    })
)

print("Depuración de variables operacionales finalizada.")
```

**Análisis de la variable temporal**

Durante el análisis exploratorio se identificó una concentración anormalmente alta de eventos registrados exactamente a las **04:00:00 UTC** y **05:00:00 UTC**. Una inspección de los registros mostró que este comportamiento se presenta principalmente en eventos históricos (décadas de 1960, 1970 y principios de 1980), lo que sugiere que dichas horas no corresponden necesariamente a la hora real de ocurrencia del accidente.

Este patrón es consistente con un proceso de estandarización de fechas, en el cual los registros que originalmente contenían únicamente la fecha fueron almacenados con una hora por defecto (00:00 hora local) y posteriormente convertidos a tiempo universal coordinado (UTC), generando valores de **04:00:00** o **05:00:00 UTC**, dependiendo del huso horario y del horario de verano.

Con el fin de evitar un sesgo en la distribución horaria de los accidentes, el análisis por hora excluye únicamente los registros cuya hora corresponde exactamente a **04:00:00 UTC** o **05:00:00 UTC**. Los registros con horas diferentes (por ejemplo, 04:17, 05:18 o 05:42) se conservan, ya que representan horas de ocurrencia registradas con mayor precisión.

Esta decisión afecta exclusivamente el análisis horario y no modifica el conjunto de datos utilizado en el resto del estudio.

```python
# ==============================================================================
# 16.2 ANÁLISIS TEMPORAL - COMPARACIÓN ENTRE TIPOS DE EVENTO
# ==============================================================================

# ------------------------------------------------------------------------------
# Configuración
# ------------------------------------------------------------------------------

orden_meses = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

orden_dias = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

COLORES = {
    "ACC": "steelblue",
    "INC": "darkorange",
    "OCC": "forestgreen"
}

# Excluir horas por defecto (04:00 y 05:00 UTC)
df_hour = df_clean[
    ~(
        (df_clean["EventDate"].dt.hour.isin([0, 4, 5, 23])) &
        (df_clean["EventDate"].dt.minute == 0) &
        (df_clean["EventDate"].dt.second == 0)
    )
]

# ------------------------------------------------------------------------------
# Figura
# ------------------------------------------------------------------------------

fig, ax = plt.subplots(2, 2, figsize=(16, 12))

# ------------------------------------------------------------------------------
# Eventos por año
# ------------------------------------------------------------------------------

tabla = pd.crosstab(
    df_clean["Year"],
    df_clean["EventType"]
)

tabla.plot(
    kind="bar",
    stacked=True,
    color=[COLORES[c] for c in tabla.columns],
    ax=ax[0,0]
)

ax[0,0].set_title("Eventos por año")
ax[0,0].set_xlabel("Año")
ax[0,0].set_ylabel("Número de eventos")

ax[0,0].set_xticks(range(0, len(tabla), 2))
ax[0,0].set_xticklabels(
    tabla.index[::2],
    rotation=90
)

# ------------------------------------------------------------------------------
# Eventos por mes
# ------------------------------------------------------------------------------

tabla = (
    pd.crosstab(
        df_clean["MonthName"],
        df_clean["EventType"]
    )
    .reindex(orden_meses)
)

tabla.plot(
    kind="bar",
    stacked=True,
    color=[COLORES[c] for c in tabla.columns],
    ax=ax[0,1]
)

ax[0,1].set_title("Eventos por mes")
ax[0,1].set_xlabel("Mes")
ax[0,1].set_ylabel("Número de eventos")

# ------------------------------------------------------------------------------
# Eventos por día de la semana
# ------------------------------------------------------------------------------

tabla = (
    pd.crosstab(
        df_clean["WeekDay"],
        df_clean["EventType"]
    )
    .reindex(orden_dias)
)

tabla.plot(
    kind="bar",
    stacked=True,
    color=[COLORES[c] for c in tabla.columns],
    ax=ax[1,0]
)

ax[1,0].set_title("Eventos por día de la semana")
ax[1,0].set_xlabel("Día")
ax[1,0].set_ylabel("Número de eventos")

# ------------------------------------------------------------------------------
# Eventos por hora
# ------------------------------------------------------------------------------

tabla = (
    pd.crosstab(
        df_hour["Hour"],
        df_hour["EventType"]
    )
    .sort_index()
)

tabla.plot(
    kind="bar",
    stacked=True,
    color=[COLORES[c] for c in tabla.columns],
    ax=ax[1,1]
)

ax[1,1].set_title("Eventos por hora")
ax[1,1].set_xlabel("Hora")
ax[1,1].set_ylabel("Número de eventos")

# ------------------------------------------------------------------------------
# Ajuste final
# ------------------------------------------------------------------------------

plt.tight_layout()
plt.show()
```

**Resultados del análisis temporal**

El análisis temporal evidencia una evolución significativa de la accidentalidad aeronáutica registrada por la NTSB a lo largo del período de estudio. Se observa un incremento sostenido de los eventos hasta finales de la década de 1960, alcanzando el máximo histórico alrededor de 1967, seguido por una disminución progresiva durante las décadas posteriores. Desde comienzos de los años 2000 la frecuencia anual presenta una tendencia relativamente estable, con valores considerablemente inferiores a los observados en las décadas anteriores.

La distribución mensual muestra un marcado comportamiento estacional. Los meses de **junio, julio y agosto** concentran el mayor número de eventos, mientras que **enero, febrero y diciembre** presentan la menor actividad. Este patrón coincide con el incremento de las operaciones aéreas durante la temporada de verano en los Estados Unidos.

Respecto a la distribución semanal, los **fines de semana**, particularmente **sábado y domingo**, registran la mayor cantidad de eventos, seguidos por el viernes. Este comportamiento es consistente con el aumento de vuelos recreativos y de aviación general durante estos días.

```python
import nltk

nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
# ==============================================================================
# 18.1 PREPARACIÓN DEL TEXTO PARA NLP
# ==============================================================================

import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# ------------------------------------------------------------------------------
# Configuración
# ------------------------------------------------------------------------------

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

# ------------------------------------------------------------------------------
# Función de limpieza
# ------------------------------------------------------------------------------

def limpiar_texto(texto):

    if pd.isna(texto):
        return ""

    # Minúsculas
    texto = texto.lower()

    # Eliminar URLs
    texto = re.sub(r"http\S+", " ", texto)

    # Eliminar números
    texto = re.sub(r"\d+", " ", texto)

    # Eliminar puntuación
    texto = re.sub(r"[^a-z\s]", " ", texto)

    # Tokenización
    palabras = texto.split()

    # Eliminar stopwords
    palabras = [
        p for p in palabras
        if p not in stop_words
    ]

    # Lematización
    palabras = [
        lemmatizer.lemmatize(p)
        for p in palabras
    ]

    return " ".join(palabras)

# ------------------------------------------------------------------------------
# Crear corpus limpio
# ------------------------------------------------------------------------------

df_clean["ProbableCause_Clean"] = (
    df_clean["ProbableCause"]
    .fillna("")
    .apply(limpiar_texto)
)

print("="*80)
print("CORPUS LIMPIO GENERADO")
print("="*80)

print(df_clean[
    ["ProbableCause","ProbableCause_Clean"]
].head(10))
```

```python
# ==============================================================================
# 18.2 PALABRAS MÁS FRECUENTES
# ==============================================================================

from collections import Counter

# ------------------------------------------------------------------------------
# Construir lista de palabras
# ------------------------------------------------------------------------------

palabras = " ".join(
    df_clean["ProbableCause_Clean"]
).split()

# Contar frecuencias
frecuencias = Counter(palabras)

# Top 50 palabras
top_palabras = (
    pd.DataFrame(
        frecuencias.most_common(50),
        columns=["Palabra", "Frecuencia"]
    )
)

print("=" * 80)
print("TOP 50 PALABRAS MÁS FRECUENTES")
print("=" * 80)

display(top_palabras)

# ------------------------------------------------------------------------------
# Gráfica
# ------------------------------------------------------------------------------

plt.figure(figsize=(12,10))

plt.barh(
    top_palabras["Palabra"][::-1],
    top_palabras["Frecuencia"][::-1]
)

plt.title("50 palabras más frecuentes en ProbableCause")
plt.xlabel("Frecuencia")
plt.ylabel("Palabra")

plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# 18.3 DEPURACIÓN DEL CORPUS PARA NLP
# ==============================================================================

# ------------------------------------------------------------------------------
# Stopwords específicas del dominio aeronáutico (NTSB)
# ------------------------------------------------------------------------------

STOPWORDS_AERO = {

    # Redacción de los informes
    "resulted",
    "resulting",
    "factor",
    "contributing",
    "due",
    "reason",
    "related",
    "could",
    "determined",

    # Palabras demasiado generales
    "accident"

}

# ------------------------------------------------------------------------------
# Función para eliminar stopwords aeronáuticas
# ------------------------------------------------------------------------------

def limpiar_corpus_nlp(texto):

    palabras = texto.split()

    palabras = [
        palabra
        for palabra in palabras
        if palabra not in STOPWORDS_AERO
    ]

    return " ".join(palabras)

# ------------------------------------------------------------------------------
# Crear nuevo corpus
# ------------------------------------------------------------------------------

df_clean["ProbableCause_NLP"] = (
    df_clean["ProbableCause_Clean"]
    .apply(limpiar_corpus_nlp)
)

print("="*80)
print("CORPUS NLP GENERADO")
print("="*80)

display(
    df_clean[
        [
            "ProbableCause",
            "ProbableCause_Clean",
            "ProbableCause_NLP"
        ]
    ].head(10)
)
```

```python
# ==============================================================================
# 19.2 CLASIFICADOR AUTOMÁTICO ADREP
# ==============================================================================

def clasificar_adrep(texto):

    if pd.isna(texto) or texto.strip() == "":
        return "UNKNOWN"

    texto = texto.lower()

    for categoria, palabras in ADREP.items():

        for patron in palabras:

            if patron in texto:
                return categoria

    return "OTHER"

# ------------------------------------------------------------------------------
# Aplicar clasificador
# ------------------------------------------------------------------------------

df_clean["ADREP"] = (
    df_clean["ProbableCause_NLP"]
    .apply(clasificar_adrep)
)

print("="*80)
print("CLASIFICACIÓN ADREP")
print("="*80)

display(
    df_clean["ADREP"]
    .value_counts()
    .to_frame("Frecuencia")
)

print()

print(
    f"Categorías encontradas: {df_clean['ADREP'].nunique()}"
)
```

```python
# ==============================================================================
# 19.4 EVALUACIÓN DE LA COBERTURA DEL CLASIFICADOR ADREP
# ==============================================================================

print("=" * 80)
print("COBERTURA DEL CLASIFICADOR ADREP")
print("=" * 80)

total = len(df_clean)

con_texto = (df_clean["ProbableCause_NLP"] != "").sum()

clasificados = (
    (df_clean["ProbableCause_NLP"] != "") &
    (df_clean["ADREP"] != "OTHER")
).sum()

print(f"Total de eventos               : {total:,}")
print(f"Eventos con causa probable     : {con_texto:,}")
print(f"Eventos clasificados           : {clasificados:,}")
print(f"Cobertura sobre textos         : {clasificados/con_texto:.2%}")
```

# **FASE VI Integración y análisis multivariable**

**Bloque 22**

Diagrama de Sankey

Flujo entre las principales variables del accidente**

Con el propósito de comprender la secuencia de eventos que conduce desde el contexto operacional hasta las consecuencias finales del accidente, se empleó un **Diagrama de Sankey**.

Esta representación permite visualizar el flujo de los eventos entre diferentes variables categóricas, donde el grosor de cada conexión es proporcional al número de accidentes que siguen una determinada trayectoria.

Para facilitar la interpretación se seleccionaron las variables que representan las diferentes etapas del evento aeronáutico:

- Condiciones meteorológicas (*Weather Condition*)
- Categoría ICAO ADREP
- Nivel de daño de la aeronave (*Aircraft Damage*)
- Nivel máximo de lesión (*Highest Injury Level*)

De esta manera es posible identificar las rutas predominantes asociadas a los accidentes e incidentes aeronáuticos registrados por la NTSB.

```python
# ==============================================================================
# 22.1 PREPARACIÓN DE LOS DATOS PARA EL DIAGRAMA DE SANKEY
# ==============================================================================

from collections import Counter

# --------------------------------------------------------------------------
# Variables seleccionadas
# --------------------------------------------------------------------------

df_sankey = df_clean[
    [
        "WeatherCondition",
        "ADREP",
        "AirCraftDamage",
        "HighestInjuryLevel"
    ]
].copy()

# Eliminar registros sin información útil

df_sankey = df_sankey[
    ~df_sankey["ADREP"].isin(["UNKNOWN","OTHER"])
]

df_sankey = df_sankey.dropna()

print("="*80)
print("DATASET PARA SANKEY")
print("="*80)

print(f"Registros: {len(df_sankey):,}")
```

```python
# ==============================================================================
# 22.2 CONSTRUCCIÓN DE LOS NODOS
# ==============================================================================

# --------------------------------------------------------------------------
# Categorías únicas
# --------------------------------------------------------------------------

weather = sorted(df_sankey["WeatherCondition"].unique())
adrep = sorted(df_sankey["ADREP"].unique())
damage = sorted(df_sankey["AirCraftDamage"].unique())
injury = sorted(df_sankey["HighestInjuryLevel"].unique())

# --------------------------------------------------------------------------
# Etiquetas visibles
# --------------------------------------------------------------------------

labels = (
    weather +
    adrep +
    damage +
    injury
)

# --------------------------------------------------------------------------
# Claves únicas para evitar categorías repetidas
# --------------------------------------------------------------------------

keys = (
    [("Weather", x) for x in weather] +
    [("ADREP", x) for x in adrep] +
    [("Damage", x) for x in damage] +
    [("Injury", x) for x in injury]
)

# --------------------------------------------------------------------------
# Diccionario nodo → índice
# --------------------------------------------------------------------------

node_dict = {
    key: i
    for i, key in enumerate(keys)
}

print("="*80)
print("NODOS DEL SANKEY")
print("="*80)

print(f"Número total de nodos: {len(labels)}")

display(
    pd.DataFrame({
        "Variable":[k[0] for k in keys],
        "Categoría":[k[1] for k in keys],
        "Índice":range(len(keys))
    })
)
```

```python
# ==============================================================================
# 22.3 CONSTRUCCIÓN DE LOS ENLACES
# ==============================================================================

source = []
target = []
value = []

# --------------------------------------------------------------------------
# Weather -> ADREP
# --------------------------------------------------------------------------

flow = (
    df_sankey
    .groupby(["WeatherCondition", "ADREP"])
    .size()
    .reset_index(name="Frecuencia")
)

for _, row in flow.iterrows():

    source.append(
        node_dict[("Weather", row["WeatherCondition"])]
    )

    target.append(
        node_dict[("ADREP", row["ADREP"])]
    )

    value.append(
        row["Frecuencia"]
    )

# --------------------------------------------------------------------------
# ADREP -> Damage
# --------------------------------------------------------------------------

flow = (
    df_sankey
    .groupby(["ADREP", "AirCraftDamage"])
    .size()
    .reset_index(name="Frecuencia")
)

for _, row in flow.iterrows():

    source.append(
        node_dict[("ADREP", row["ADREP"])]
    )

    target.append(
        node_dict[("Damage", row["AirCraftDamage"])]
    )

    value.append(
        row["Frecuencia"]
    )

# --------------------------------------------------------------------------
# Damage -> Injury
# --------------------------------------------------------------------------

flow = (
    df_sankey
    .groupby(["AirCraftDamage", "HighestInjuryLevel"])
    .size()
    .reset_index(name="Frecuencia")
)

for _, row in flow.iterrows():

    source.append(
        node_dict[("Damage", row["AirCraftDamage"])]
    )

    target.append(
        node_dict[("Injury", row["HighestInjuryLevel"])]
    )

    value.append(
        row["Frecuencia"]
    )

print("="*80)
print("ENLACES DEL SANKEY")
print("="*80)

print(f"Número de enlaces: {len(source):,}")

display(
    pd.DataFrame({
        "Origen": source,
        "Destino": target,
        "Peso": value
    }).head(20)
)
```

```python
# ==============================================================================
# 22.4 DIAGRAMA DE SANKEY
# ==============================================================================

import plotly.graph_objects as go
import numpy as np

# ------------------------------------------------------------------------------
# Seleccionar Top 8 categorías ADREP
# ------------------------------------------------------------------------------

TOP_ADREP = (
    df_clean["ADREP"]
    .value_counts()
    .head(8)
    .index
)

# ------------------------------------------------------------------------------
# Preparación de los datos
# ------------------------------------------------------------------------------

df_sankey = df_clean[
    [
        "WeatherCondition",
        "ADREP",
        "AirCraftDamage",
        "HighestInjuryLevel"
    ]
].copy()

# Agrupar categorías poco frecuentes

df_sankey["ADREP"] = np.where(
    df_sankey["ADREP"].isin(TOP_ADREP),
    df_sankey["ADREP"],
    "Categorías menores"
)

# Eliminar únicamente categorías desconocidas

df_sankey = df_sankey[
    (df_sankey["ADREP"] != "UNKNOWN") &
    (df_sankey["WeatherCondition"] != "UNKNOWN") &
    (df_sankey["AirCraftDamage"] != "UNKNOWN")
]

df_sankey = df_sankey.dropna()

# ------------------------------------------------------------------------------
# Definición de nodos
# ------------------------------------------------------------------------------

weather = ["VMC", "IMC"]

adrep = (
    df_sankey["ADREP"]
    .value_counts()
    .index
    .tolist()
)

damage = [
    "SUBSTANTIAL",
    "DESTROYED",
    "MINOR",
    "NONE"
]

injury = [
    "Fatal",
    "Serious",
    "Minor"
]

labels = weather + adrep + damage + injury

keys = (
    [("Weather", x) for x in weather] +
    [("ADREP", x) for x in adrep] +
    [("Damage", x) for x in damage] +
    [("Injury", x) for x in injury]
)

node_dict = {
    k: i
    for i, k in enumerate(keys)
}

# ------------------------------------------------------------------------------
# Construcción de enlaces
# ------------------------------------------------------------------------------

MIN_FLUJO = 20

source = []
target = []
value = []
link_color = []

COLOR_WEATHER = "rgba(65,105,225,0.45)"
COLOR_ADREP   = "rgba(255,140,0,0.45)"
COLOR_DAMAGE  = "rgba(178,34,34,0.45)"

# ------------------------------------------------------------------------------
# Weather → ADREP
# ------------------------------------------------------------------------------

flow = (
    df_sankey
    .groupby(["WeatherCondition","ADREP"])
    .size()
    .reset_index(name="Frecuencia")
)

flow = flow[flow["Frecuencia"] >= MIN_FLUJO]

for _, row in flow.iterrows():

    source.append(node_dict[("Weather", row["WeatherCondition"])])
    target.append(node_dict[("ADREP", row["ADREP"])])
    value.append(row["Frecuencia"])
    link_color.append(COLOR_WEATHER)

# ------------------------------------------------------------------------------
# ADREP → Damage
# ------------------------------------------------------------------------------

flow = (
    df_sankey
    .groupby(["ADREP","AirCraftDamage"])
    .size()
    .reset_index(name="Frecuencia")
)

flow = flow[flow["Frecuencia"] >= MIN_FLUJO]

for _, row in flow.iterrows():

    source.append(node_dict[("ADREP", row["ADREP"])])
    target.append(node_dict[("Damage", row["AirCraftDamage"])])
    value.append(row["Frecuencia"])
    link_color.append(COLOR_ADREP)

# ------------------------------------------------------------------------------
# Damage → Injury
# ------------------------------------------------------------------------------

flow = (
    df_sankey
    .groupby(["AirCraftDamage","HighestInjuryLevel"])
    .size()
    .reset_index(name="Frecuencia")
)

flow = flow[flow["Frecuencia"] >= MIN_FLUJO]

for _, row in flow.iterrows():

    source.append(node_dict[("Damage", row["AirCraftDamage"])])
    target.append(node_dict[("Injury", row["HighestInjuryLevel"])])
    value.append(row["Frecuencia"])
    link_color.append(COLOR_DAMAGE)

# ------------------------------------------------------------------------------
# Colores de los nodos
# ------------------------------------------------------------------------------

node_colors = (

    ["royalblue"] * len(weather) +

    ["darkorange"] * len(adrep) +

    ["firebrick"] * len(damage) +

    ["forestgreen"] * len(injury)

)

# ------------------------------------------------------------------------------
# Figura
# ------------------------------------------------------------------------------

fig = go.Figure(

    go.Sankey(

        arrangement="snap",

        node=dict(

            pad=22,
            thickness=24,

            line=dict(
                color="black",
                width=0.5
            ),

            label=labels,
            color=node_colors

        ),

        link=dict(

            source=source,
            target=target,
            value=value,
            color=link_color

        )

    )

)

fig.update_layout(

    title="Flujo predominante entre las condiciones meteorológicas, la clasificación ICAO ADREP, el daño de la aeronave y el nivel máximo de lesión",

    font=dict(size=14),

    width=1550,
    height=780

)

fig.show()
```

**Resultados**

Una vez construidos los nodos y enlaces, se generó un diagrama de Sankey utilizando la biblioteca **Plotly**. Con el fin de mejorar la legibilidad de la visualización, se eliminaron los enlaces cuya frecuencia fue inferior a un umbral mínimo, conservando únicamente los flujos más representativos del conjunto de datos.

El grosor de cada conexión es proporcional al número de eventos que siguen la transición entre categorías consecutivas, permitiendo identificar de forma intuitiva los mecanismos predominantes desde las condiciones meteorológicas hasta las consecuencias finales del evento.

# FASE VII. Priorización del riesgo operacional

**Bloque 23**

Índice de Severidad Operacional

Con el fin de priorizar las categorías de la taxonomía ICAO ADREP desde la perspectiva de la seguridad operacional, se construyeron indicadores de severidad basados en las consecuencias de cada mecanismo de ocurrencia. A diferencia del análisis descriptivo realizado anteriormente, esta etapa busca identificar no solo las categorías más frecuentes, sino también aquellas que producen las consecuencias más graves en términos de pérdida de vidas humanas y daño a la aeronave.

Para ello, inicialmente se calcularon, para cada categoría ICAO ADREP, el número total de eventos, la proporción de accidentes fatales, la proporción de aeronaves destruidas y la proporción de eventos con lesiones graves. Estos indicadores constituyen la base para la construcción posterior de un índice de severidad operacional y una matriz de priorización del riesgo.

```python
# ==============================================================================
# 23.1 CONSTRUCCIÓN DE LOS INDICADORES DE SEVERIDAD
# ==============================================================================

# ------------------------------------------------------------------------------
# Registros clasificados
# ------------------------------------------------------------------------------

df_risk = df_clean[
    ~df_clean["ADREP"].isin(["UNKNOWN"])
].copy()

# ------------------------------------------------------------------------------
# Total de eventos
# ------------------------------------------------------------------------------

eventos = (
    df_risk
    .groupby("ADREP")
    .size()
    .rename("Eventos")
)

# ------------------------------------------------------------------------------
# Conteo de accidentes fatales
# ------------------------------------------------------------------------------

fatal = (
    df_risk[
        df_risk["HighestInjuryLevel"] == "Fatal"
    ]
    .groupby("ADREP")
    .size()
    .rename("Fatal")
)

# ------------------------------------------------------------------------------
# Conteo de lesiones graves
# ------------------------------------------------------------------------------

serious = (
    df_risk[
        df_risk["HighestInjuryLevel"] == "Serious"
    ]
    .groupby("ADREP")
    .size()
    .rename("Serious")
)

# ------------------------------------------------------------------------------
# Conteo de aeronaves destruidas
# ------------------------------------------------------------------------------

destroyed = (
    df_risk[
        df_risk["AirCraftDamage"] == "DESTROYED"
    ]
    .groupby("ADREP")
    .size()
    .rename("Destroyed")
)

# ------------------------------------------------------------------------------
# Unir tablas
# ------------------------------------------------------------------------------

tabla_riesgo = pd.concat(
    [
        eventos,
        fatal,
        serious,
        destroyed
    ],
    axis=1
).fillna(0)

# Convertir conteos a enteros

tabla_riesgo[
    ["Fatal","Serious","Destroyed"]
] = (
    tabla_riesgo[
        ["Fatal","Serious","Destroyed"]
    ]
    .astype(int)
)

# ------------------------------------------------------------------------------
# Calcular porcentajes respecto al total de eventos
# ------------------------------------------------------------------------------

tabla_riesgo["% Fatal"] = (
    tabla_riesgo["Fatal"] /
    tabla_riesgo["Eventos"] * 100
).round(2)

tabla_riesgo["% Serious"] = (
    tabla_riesgo["Serious"] /
    tabla_riesgo["Eventos"] * 100
).round(2)

tabla_riesgo["% Destroyed"] = (
    tabla_riesgo["Destroyed"] /
    tabla_riesgo["Eventos"] * 100
).round(2)

# ------------------------------------------------------------------------------
# Ordenar
# ------------------------------------------------------------------------------

tabla_riesgo = tabla_riesgo.sort_values(
    "Eventos",
    ascending=False
)

print("="*80)
print("INDICADORES DE SEVERIDAD OPERACIONAL")
print("="*80)

display(tabla_riesgo)
```

**Análisis de los indicadores de severidad operacional**

Los indicadores obtenidos permiten caracterizar la severidad de cada categoría de la taxonomía ICAO ADREP desde diferentes perspectivas, considerando tanto las consecuencias sobre las personas como los daños ocasionados a la aeronave. A diferencia de un único índice compuesto, el análisis individual de estos indicadores conserva la información específica de cada dimensión de la severidad y evita introducir ponderaciones subjetivas que puedan alterar la interpretación de los resultados.

En esta etapa se identifican las categorías con mayor proporción de accidentes fatales, lesiones graves y destrucción de aeronaves, proporcionando una visión integral del impacto operacional asociado a cada mecanismo de ocurrencia. Estos resultados constituyen la base para la construcción posterior de una matriz de priorización del riesgo, en la cual se combinarán la frecuencia de ocurrencia y la severidad de las consecuencias para identificar las categorías que requieren mayor atención desde la perspectiva de la seguridad operacional.

```python
# ==============================================================================
# 23.2 ANÁLISIS DE LOS INDICADORES DE SEVERIDAD OPERACIONAL
# ==============================================================================

print("="*80)
print("RANKING DE LOS PRINCIPALES INDICADORES DE SEVERIDAD")
print("="*80)

# ------------------------------------------------------------------------------
# Construcción de la tabla resumen
# ------------------------------------------------------------------------------

ranking = pd.DataFrame({

    "Ranking": range(1,6),

    "Accidentes fatales": [

        f"{idx} ({val:.2f}%)"

        for idx, val in
        tabla_riesgo["% Fatal"]
        .sort_values(ascending=False)
        .head(5)
        .items()

    ],

    "Lesiones graves": [

        f"{idx} ({val:.2f}%)"

        for idx, val in
        tabla_riesgo["% Serious"]
        .sort_values(ascending=False)
        .head(5)
        .items()

    ],

    "Aeronaves destruidas": [

        f"{idx} ({val:.2f}%)"

        for idx, val in
        tabla_riesgo["% Destroyed"]
        .sort_values(ascending=False)
        .head(5)
        .items()

    ],

    "Frecuencia": [

        f"{idx} ({val:,})"

        for idx, val in
        tabla_riesgo["Eventos"]
        .sort_values(ascending=False)
        .head(5)
        .items()

    ]

})

display(ranking)
```

**Resultados**

**Matriz Frecuencia–Severidad de las categorías ICAO ADREP**

Con el propósito de establecer prioridades desde la perspectiva de la seguridad operacional, se construyó una matriz Frecuencia–Severidad para las categorías ICAO ADREP. Esta herramienta permite identificar simultáneamente la ocurrencia de cada mecanismo de accidente y la gravedad relativa de sus consecuencias.

La frecuencia se definió como el número total de eventos registrados para cada categoría ADREP. Por su parte, la severidad se estimó mediante un indicador compuesto construido a partir de tres consecuencias operacionales de cada categoría:

- porcentaje de accidentes fatales;
- porcentaje de aeronaves destruidas;
- porcentaje de eventos con lesiones graves.

El indicador de severidad se calculó como el promedio aritmético de estos tres porcentajes, asignando inicialmente el mismo peso a cada componente al encontrarse expresados sobre una escala común (0–100 %). Esta aproximación proporciona una medida relativa de severidad que facilita la comparación entre categorías, sin introducir ponderaciones subjetivas.

Para clasificar las categorías se dividió el plano utilizando como referencia la media del número de eventos y la media del indicador de severidad, generando cuatro cuadrantes:

- **I. Alta frecuencia – Alta severidad:** categorías críticas que constituyen la mayor prioridad para la gestión del riesgo.
- **II. Baja frecuencia – Alta severidad:** eventos poco frecuentes pero con consecuencias potencialmente catastróficas.
- **III. Alta frecuencia – Baja severidad:** eventos recurrentes cuyo impacto individual es reducido, aunque representan una importante carga operacional.
- **IV. Baja frecuencia – Baja severidad:** categorías de menor prioridad relativa.

```python
# ==============================================================================
# 23.3 MATRIZ DE PRIORIZACIÓN DEL RIESGO OPERACIONAL
# Categorías ICAO ADREP
# ==============================================================================

import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

tabla_plot = tabla_riesgo.copy()

tabla_plot["Severidad"] = (
    tabla_plot[["% Fatal","% Serious","% Destroyed"]].mean(axis=1)
)

media_eventos = tabla_plot["Eventos"].mean()
media_severidad = tabla_plot["Severidad"].mean()

def clasificar(f,s):
    if f >= media_eventos and s >= media_severidad:
        return "I"
    elif f < media_eventos and s >= media_severidad:
        return "II"
    elif f >= media_eventos and s < media_severidad:
        return "III"
    return "IV"

tabla_plot["Cuadrante"]=[
    clasificar(f,s)
    for f,s in zip(tabla_plot["Eventos"],tabla_plot["Severidad"])
]

colores={
"I":"#d73027",
"II":"#fc8d59",
"III":"#4575b4",
"IV":"#91cf60"
}

fig,ax=plt.subplots(figsize=(14,9))

xmax=tabla_plot["Eventos"].max()*1.05
ymax=tabla_plot["Severidad"].max()*1.08

ax.fill_between([0,media_eventos],media_severidad,ymax,color="#fddbc7",alpha=0.18)
ax.fill_between([media_eventos,xmax],media_severidad,ymax,color="#f4cccc",alpha=0.18)
ax.fill_between([0,media_eventos],0,media_severidad,color="#d9ead3",alpha=0.18)
ax.fill_between([media_eventos,xmax],0,media_severidad,color="#d9e2f3",alpha=0.18)

ax.axvline(media_eventos,color="dimgray",ls="--",lw=2)
ax.axhline(media_severidad,color="dimgray",ls="--",lw=2)

for q,g in tabla_plot.groupby("Cuadrante"):
    ax.scatter(g["Eventos"],g["Severidad"],s=260,color=colores[q],
               edgecolor="black",linewidth=0.8,zorder=3)

offsets={
"LOC-I":(-1200,0.4),
"SCF-PP":(180,0.3),
"LOC-G":(180,-0.2),
"OTHER":(150,0.3),
"CFIT":(180,0.4),
"ARC":(100,-0.2)
}

for cat,row in tabla_plot.iterrows():
    dx,dy=offsets.get(cat,(90,0.2))
    ax.text(row["Eventos"]+dx,row["Severidad"]+dy,cat,
            fontsize=11,fontweight="bold",
            bbox=dict(boxstyle="round,pad=0.18",
                      facecolor="white",
                      edgecolor="lightgray",
                      alpha=0.9))

ax.text(media_eventos*0.18,ymax*0.88,"II",fontsize=22,fontweight="bold",alpha=.7)
ax.text(media_eventos*1.03,ymax*0.88,"I",fontsize=22,fontweight="bold",alpha=.7)
ax.text(media_eventos*1.03,ymax*0.18,"III",fontsize=22,fontweight="bold",alpha=.7)
ax.text(media_eventos*0.18,ymax*0.18,"IV",fontsize=22,fontweight="bold",alpha=.7)

handles=[
Line2D([0],[0],marker='o',color='w',markerfacecolor=colores["I"],markeredgecolor='black',markersize=10,label="I Alta frecuencia - Alta severidad"),
Line2D([0],[0],marker='o',color='w',markerfacecolor=colores["II"],markeredgecolor='black',markersize=10,label="II Baja frecuencia - Alta severidad"),
Line2D([0],[0],marker='o',color='w',markerfacecolor=colores["III"],markeredgecolor='black',markersize=10,label="III Alta frecuencia - Baja severidad"),
Line2D([0],[0],marker='o',color='w',markerfacecolor=colores["IV"],markeredgecolor='black',markersize=10,label="IV Baja frecuencia - Baja severidad")
]
ax.legend(handles=handles,loc="lower right",title="Cuadrantes")

ax.set_xlim(-300,xmax)
ax.set_ylim(-1,ymax)
ax.set_xlabel("Número de eventos",fontsize=14,fontweight="bold")
ax.set_ylabel("Indicador compuesto de severidad (%)",fontsize=14,fontweight="bold")
ax.set_title("Matriz Frecuencia–Severidad de las categorías ICAO ADREP",
             fontsize=18,fontweight="bold")
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

**Resultados de la matriz Frecuencia–Severidad**

La matriz permitió identificar las categorías ICAO ADREP que representan la mayor prioridad desde la perspectiva de la seguridad operacional al combinar simultáneamente la frecuencia de ocurrencia y la gravedad de sus consecuencias.

El cuadrante I (alta frecuencia–alta severidad) estuvo dominado por **LOC-I (Loss of Control In-flight)**, la categoría con mayor número de eventos y uno de los indicadores de severidad más elevados, seguida por **CFIT (Controlled Flight Into Terrain)** y la categoría **OTHER**, evidenciando que estos mecanismos concentran una proporción importante del riesgo operacional.

En el cuadrante II (baja frecuencia–alta severidad) se ubicaron categorías como **TURB**, **NAV**, **CTOL**, **CABIN**, **ICE**, **WSTRW** y **ATM**. Aunque presentan un número reducido de eventos, sus consecuencias son considerablemente más severas, por lo que constituyen riesgos de baja probabilidad pero alto impacto.

El cuadrante III (alta frecuencia–baja severidad) agrupó principalmente a **SCF-PP**, **SCF-NP**, **LOC-G** y **ARC**, categorías caracterizadas por una elevada recurrencia, pero con consecuencias relativamente menos graves en comparación con las categorías críticas.

Finalmente, el cuadrante IV (baja frecuencia–baja severidad) incluyó categorías como **FUEL**, **RE** y **RAMP**, las cuales presentan tanto baja ocurrencia como menor impacto relativo sobre la seguridad operacional.

En conjunto, la matriz proporciona una herramienta de priorización que permite diferenciar entre problemas altamente recurrentes y aquellos que, aun siendo menos frecuentes, generan consecuencias significativamente más severas, facilitando la definición de estrategias de mitigación orientadas al riesgo.
